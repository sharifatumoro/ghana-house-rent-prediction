```python
cor = df[['price','bathrooms','bedrooms','floor_area','lat','lng','parking_space']].corr()
cor
```

```python
plt.Figure(figsize=(15,10))
sns.heatmap(cor, annot=True, cmap='coolwarm')
plt.title("Correlation Map")
```

Bedrooms and  bathrooms have a moderate positive correlation with price meaning the higher the number of bedrooms and bathrooms, the higher the price.Latitude and longitude have a weak negative and a weak positive correlation respectively with price and will be dropped subsequently.The floor area of the property and the availability of parking space has no correlation with price but will be maintained because intuitively these variables affect price. This suggest that their relationship with price may not be a linear one. 
Interestinly, the relationship between bathrooms and bedrooms is a strong positive one which can  cause multicollinearity. So multicollinearity will be checked for.

```python
sns.boxplot(x='region',y='price',data=df)
plt.xticks(rotation=90)
plt.title("Distribution of Prices across Regions")
plt.show()
```

It can be observed that prices is Greater Accra vary significantly as it recorded the highest price and has the widest spread.It median rent is higher than most of the regions and the presence of many outliers suggest very expensive houses.
The volta Region has the highest median rent and a wider spread indicating the presence of both moderate and very expensive houses.
All the other regions have lower spread and and very few expensive houses indicating that rent in these regions are less expensive. 

```python
sns.boxplot(x='category',y='price',data=df)
plt.xticks(rotation=90)
plt.title("Distribution of Prices across Categories")
plt.show()
```

Mansions and Townhouses have the widest spread and the highest median rent suggesting that average rent price for these two are higher than the others. Detached houses have higher median rent compared to Flats and semi-detached houses.The presence of outliers suggest that some detached hoouses are really expensive to rent. Flats are the second to least with lower spread and lower median rent.It has the most outliers indicating that most Flats have higher rent prices compared to average.This might be influenced by location or the locality.

```python
sns.boxplot(x='is_furnished',y='price',data=df)
plt.xticks(rotation=0)
plt.title("Distribution of Prices across Property's Nature")
plt.show()
```

The nature of a property being furnished,semi-furnished or unfurnished has a great influence on it price. Furnished houses have wider spread and the highest median with fewer outliers suggesting that furnished houses are the most expensive to rent. This is followed by semi-furnished houses with wide spread and a relatively lower median rent with many outliers indicating that though rent prices for semi-furnished properties are less expensive on average, some are really expensive. Unfurnished houses have the least prices on average but the presence of dense outliers suggest that most of their prices go beyond the average and this might be influenced by a lot of factors including locality,region,location,number of bedrooms etc.
