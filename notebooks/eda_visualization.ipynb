### Outliers detection

### Remove Outliers

```python
numerical_col = ['price','bathrooms','bedrooms','floor_area']

for col in numerical_col:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 -Q1
    lower = Q1 -1.5 * IQR
    upper = Q3 + 1.5 * IQR
    df = df[(df[col] >= lower ) & (df[col] <= upper )]

```

### Visualization after removing outliers

```python
sns.boxplot(y='price',data=df)
plt.title("Boxplot of price")
plt.show()
```

## Encoding categorical values to numeric

### Encoding condition to numerical values using ordinal encoding

```python
df['condition'].unique()
```

```python
condition_order = {'Used': 0, 'Renovated': 1, 'New': 2}
```

```python
df['condition_encoded'] = df['condition'].map(condition_order)
```

```python
df['condition'] = df['condition_encoded']
df.drop('condition_encoded', axis=1, inplace=True)
```

```python
df[['condition']].head()
```

### Encoding is_furnished to numerical values using ordinal encoding

```python
df['is_furnished'].value_counts()
```

```python
furnished_order = {'Unfurnished': 0, 'Semi-Furnished': 1, 'Furnished': 2}
```

```python
df['furnished_encoded'] = df['is_furnished'].map(furnished_order)
```

```python
df['furnished'] = df['furnished_encoded']
df.drop(['furnished_encoded','is_furnished'], axis=1, inplace=True)
```

```python
df.head()
```

Target mean encoding will be used for the rest  of the categorical variables in model_training.ipynb.

## Drop irrelevant columns

```python
df.drop(['floor_area','lat','lng','parking_space'], axis=1, inplace = True)
```
